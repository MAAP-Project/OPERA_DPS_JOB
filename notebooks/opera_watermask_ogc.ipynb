```python
# Instantiate maap-py to the UAT environment
from maap.maap import MAAP
maap = MAAP(maap_host='api.uat.maap-project.org')
```

```python
# Determine resource queues available to user
data = maap.get_queues().json()
print(data["queues"])
```

```python
# Submit job
response = maap.submit_job(tag="test-job", # Tag to easily search for your job in the View Jobs view
    process_id=60, # Unique ID belonging to the process you want to execute (operawatermask1:ogc)
    queue="maap-dps-sandbox", # Replace this with one of the resource queues available to you
    inputs={
      "SHORT_NAME":"OPERA_L3_DISP-S1_V1",
      "TEMPORAL":"2016-07-01T00:00:00Z,2024-12-31T23:59:59Z",
      "IDX_WINDOW":"0:1024,0:1024"
    })
print(response.json())
```
